```python
def coupling_indices_real(lmax):
    for l in range(lmax + 1):
        for m in range(-l, l + 1):
            for lamda in range(lmax + 1):
                for mu in range(-lamda, lamda + 1):
                    # triangle rule
                    if lamda == l + 1 or lamda == l - 1:
                        # magnetic selection rule for real harmonics
                        if (m == mu or m == mu + 1 or m == mu - 1 or 
                            m == -mu or m == -mu + 1 or m == -mu - 1):
                            yield f"C_{{{l}{m},{lamda}{mu}}} \\cdot \\nabla"
                    # diagonal term
                    elif lamda == l and mu == m:
                        yield f"\\sigma^t_g"
                    else:
                        yield "0"

f = coupling_indices_real(3)

                    
```

```python
def latex_real_coupling_matrix_vdots(lmax):
    # Build the list of real spherical harmonic labels
    labels = []
    for l in range(lmax + 1):
        for m in range(-l, l + 1):
            labels.append((l, m))
    
    n = len(labels)
    latex = "\t\\tiny\n\t\\begin{bmatrix}\n\t"
    
    for i, (l_row, m_row) in enumerate(labels):
        row_entries = []
        for j, (l_col, m_col) in enumerate(labels):
            if l_col == l_row and m_col == m_row:
                row_entries.append("\\sigma^t_g")
            elif l_col == l_row + 1 or l_col == l_row - 1:
                if (m_row == m_col or m_row == m_col + 1 or m_row == m_col - 1 or
                    m_row == -m_col or m_row == -m_col + 1 or m_row == -m_col - 1):
                    row_entries.append(f"C_{{{l_row}{m_row},{l_col}{m_col}}} \\cdot \\nabla")
                else:
                    row_entries.append("0")
            else:
                row_entries.append("0")
        # Append \cdots at the end of the row
        row_entries.append("\\cdots")
        latex += " & ".join(row_entries) + " \\\\\n\t"
    
    # Append \vdots row at the bottom
    latex += " \\vdots" + " & \\vdots" * n + " & \\ddots \\\\\n\t"
    
    latex += "\\end{bmatrix}\n"
    return latex

# Example usage for lmax=3
latex_code = latex_real_coupling_matrix_vdots(3)
print(latex_code)
```

```python
def coupling_indices(Lmax):
    rows = []
    for l in range(Lmax+1):
        for m in range(-l,l+1):

            for dl in (-1,1):
                lam = l+dl
                if lam<0 or lam>Lmax:
                    continue

                for dmu in (-1,0,1):
                    mu = m+dmu
                    if abs(mu)>lam:
                        continue

                    rows.append((l,m,lam,mu))

    return jnp.array(rows)      # [nnz,4]

coupling_indices(3)
```

```python
def lm_index(l, m):
    return l*l + l + m   # compact lexicographic
```

```python
def w3j_l_l1(l, lam, m, mu, M):
    # only valid for lam=l±1
    phase = (-1.0)**(l-m)

    if lam == l+1:
        num = (l+1+m+M)*(l+1-m-M)
        den = (l+1)*(2*l+1)*(2*l+3)
    else:  # lam=l-1
        num = (l-m+M)*(l+m-M)
        den = l*(2*l-1)*(2*l+1)

    return phase * jnp.sqrt(jnp.maximum(num,0)/den)

def build_C_exact(Lmax):

    pairs = []
    values = []

    for l in range(Lmax+1):
        for m in range(-l,l+1):

            i = lm_index(l,m)

            for dl in (-1,1):
                lam = l+dl
                if lam<0 or lam>Lmax:
                    continue

                for dmu in (-1,0,1):
                    mu = m+dmu
                    if abs(mu)>lam:
                        continue

                    j = lm_index(lam,mu)

                    Cvec = jnp.zeros(3)

                    for M in (-1,0,1):
                        if m+mu+M!=0:
                            continue

                        w = w3j_l_l1(l,lam,m,mu,M)

                        # Ω components in complex basis
                        if M==0:
                            vec = jnp.array([0.,0., jnp.sqrt(4*jnp.pi/3)])
                        elif M==1:
                            vec = jnp.array([-jnp.sqrt(2*jnp.pi/3),0.,0.])
                        else:
                            vec = jnp.array([jnp.sqrt(2*jnp.pi/3),0.,0.])

                        Cvec = Cvec + w*vec

                    if jnp.linalg.norm(Cvec)>0:
                        pairs.append((i,j))
                        values.append(Cvec)

    return jnp.array(pairs), jnp.array(values)


build_C_exact(3)
```
