```python
theta_now = Aop.transport_operator.solution_domain_data.domain.cross_sections_indexed.cross_section_data.total[:,0]
```

```python
def parametrise_A(theta, Aop, rhs):
    return eqx.tree_at(lambda Aop : Aop.transport_operator.solution_domain_data.domain.cross_sections_indexed.cross_section_data.total, Aop, replace_fn = lambda total : total.at[:,0].set(theta))


@eqx.filter_jit
def solve_and_sum(theta, Aop, rhs):
    Aop_param = parametrise_A(theta, Aop, rhs)
    bicgstab_solver = lx.BiCGStab(1e-10, 1e-10, max_steps = 20)
    solution = lx.linear_solve(Aop_param, rhs, bicgstab_solver)
    return jnp.sum(solution.value)

```

```python
def total_fds(dx, theta, Aop, rhs):
    result = jnp.zeros_like(theta)
    for i in range(len(theta)):
         theta_perturbed = theta.at[i].add(dx)
         a0 = solve_and_sum(theta, Aop, rhs)
         adx = solve_and_sum(theta_perturbed, Aop, rhs)
         anumerical_grad_i = (adx - a0) / dx
         result = result.at[i].set(anumerical_grad_i)
    return result
total_fds(1e-6, theta_now, Aop_new, rhs)
```
